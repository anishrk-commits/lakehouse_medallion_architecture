```python
query = """
SELECT 
    ROW_NUMBER() OVER (ORDER BY product_id) as product_number,
    p.product_id,
    p.product_key,
    p.product_name,
    p.product_line,
    pc.category,
    pc.subcategory,
    p.product_cost,
    p.product_start_date,
    p.product_end_date,
    pc.maintenance_flag

FROM workspace.silver.crm_products p
LEFT JOIN workspace.silver.erp_product_category pc
    ON p.product_key LIKE CONCAT(pc.category_id, '%') AND p.product_end_date IS NOT NULL
"""

df = spark.sql(query)
```

```python
df.write.mode("overwrite").format("delta").saveAsTable("workspace.gold.dim_products")
```
