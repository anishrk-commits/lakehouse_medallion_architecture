```python
query = """
SELECT 
cc.customer_id,
cc.customer_key,
cc.first_name,
cc.last_name,
cc.marital_status,
CASE
    WHEN cc.gender <> 'n/a' THEN cc.gender
    ELSE COALESCE(cc.gender, ec.gender, 'n/a') END
     as gender,
cc.created_date,
cl.country,
ec.birth_date
FROM workspace.silver.crm_customers cc
LEFT JOIN workspace.silver.erp_customer_location cl ON cc.customer_key = cl.customer_key
LEFT JOIN workspace.silver.erp_customers ec ON ec.customer_id = cc.customer_key
"""
df = spark.sql(query)
```

```python
df.write.mode("overwrite").format("delta").saveAsTable("workspace.gold.dim_customers")
```
