### Data Cleaning & Transformations

Duplicate customer rows

```python
%sql
SELECT cst_id, COUNT(cst_id)
FROM workspace.bronze.crm_cust_info
GROUP BY cst_id
HAVING COUNT(cst_id) > 1;

```

Martial status has nulls

```python
%sql

SELECT DISTINCT cst_marital_status
FROM workspace.bronze.crm_cust_info
```

Gender has nulls

```python
%sql
SELECT DISTINCT cst_gndr
FROM workspace.bronze.crm_cust_info
```

First name text validation

```python
%sql
SELECT *
FROM workspace.bronze.crm_cust_info
WHERE cst_firstname <> TRIM(cst_firstname)
```

Last name text validation

```python
%sql
SELECT *
FROM workspace.bronze.crm_cust_info
WHERE cst_lastname <> TRIM(cst_lastname)
```
