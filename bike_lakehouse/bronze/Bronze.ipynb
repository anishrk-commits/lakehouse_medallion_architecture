```python
folder_path = "/Volumes/workspace/bronze/raw_sources"

files = dbutils.fs.ls(folder_path)

for file in files:
    table_name = file.name[:-4]
    df = spark.read.format("csv").option("header", "true").option("inferSchema", "true").load(file.path)
    if table_name[0].isupper():
        df.write.mode("overwrite").saveAsTable("bronze." + "erp_"+table_name)
    else:
        df.write.mode("overwrite").saveAsTable("bronze." + "crm_"+table_name)
```
